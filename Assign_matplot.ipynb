```python
import matplotlib.pyplot as plt
class ViratKohliIPL:
    def __init__(self):
        self.years = list(range(2008, 2026))
        self.runs = [165, 246, 307, 557, 364, 634, 359, 505, 973,
            308, 530, 464, 466, 341, 639, 730, 542, 615 ]
        self.averages = [15.0, 22.3, 27.9, 46.4, 28.0, 45.3, 27.6, 45.9, 81.1,
            22.4, 53.0, 33.1, 42.4, 28.4, 53.2, 60.8, 48.3, 51.2]
    def plot_performance(self):
        fig, axs = plt.subplots(2, 1, figsize=(10, 8))
        axs[0].plot(self.years, self.runs, marker='o')
        axs[0].set_title("Virat Kohli IPL Runs (2008–2025)")
        axs[0].set_xlabel("Year")
        axs[0].set_ylabel("Runs Scored")
        axs[0].grid(True)
        axs[1].plot(self.years, self.averages, marker='s')
        axs[1].set_title("Virat Kohli IPL Batting Average (2008–2025)")
        axs[1].set_xlabel("Year")
        axs[1].set_ylabel("Batting Average")
        axs[1].grid(True)

        plt.tight_layout()
        plt.show()
kohli = ViratKohliIPL()
kohli.plot_performance()
```
