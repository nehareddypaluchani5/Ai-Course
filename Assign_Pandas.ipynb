<h1>Part 3: Data Cleaning </h1>

```python
df.drop(columns=['S N'],inplace =True)
```

```python
df['Inches'].fillna(df['Inches'].mode()[0],inplace=True)
```

```python
df['Ram'].fillna(df['Ram'].mode()[0],inplace=True)
```

```python
df.info()
```

```python
df['Inches'] = pd.to_numeric(df['Inches'],errors='coerce')
```

```python
df['Ram'] = df['Ram'].astype(str).str.replace('GB','', regex=False)
df['Ram'] = pd.to_numeric(df['Ram'], errors='coerce')
df['Ram'].fillna(df['Ram'].mode()[0], inplace=True)
```

```python
df['Weight'] = df['Weight'].astype(str).str.replace('kg','', regex=False)
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')
df['Weight'].fillna(df['Weight'].mode()[0], inplace=True)
```
