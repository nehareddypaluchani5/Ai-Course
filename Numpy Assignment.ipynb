<h4> <b> Statistical Operations </b> </h4>

```python
# Creating a New Array
arr = np.array([[ 2,  3,  4],
       [ 4,  3, 17],
       [89, 33, 12]])
arr
```

```python
# Sum Operations
# Produce total sum, row wise sum (first elements), columns wise sum
total_sum = np.sum(arr)
print(total_sum)

```

```python
column_sum = np.sum(arr, axis=0)
column_sum
```

```python
row_sum = np.sum(arr, axis=1)
row_sum
```
