<h1>Question 1</h1>

```python
employee_name = input("enter a name :")
monthly_salary = int(input("enter a salary :"))
experience_years = int(input("enter a experience :"))
def calucate_bonus(salary,experience):
    if experience >=5 and salary<50000 :
        bonus = salary+0.10
    elif 3>= experience <5 :
        bonus =salary+0.5
    else  :
        bonus =0
    return bonus
bonus_salary =calucate_bonus(monthly_salary , experience_years)
final_salary = monthly_salary+bonus_salary
print("Employee Name :   ",employee_name)
print("Experience :   ",experience_years)
print("Bonus Amount : RS ",bonus_salary)
print("Final Salary : Rs",final_salary)
```

<h2>Question 2</h2>

```python
student_name=input("enter a student name :")
total_classes =int(input("enter the total classes :  "))
attendend_classes=int(input("enter the attend classes :"))
marks=[]
for i in range(1,4):
    mark=int(input(f"enter the subject marks{i}:"))
    marks.append(mark)
def check_eligibility(Attendance_percentage,marks_list):
    if Attendance_percentage >=75 and all (mark >=40 for mark in marks_list) :
        return "Eligibity"
    else:
        return "Not Eligibility"
Attendance_Percentage=(total_classes/attendend_classes )*100
Eligibility_status =check_eligibility(Attendance_Percentage,marks)
print("enter a Student name:  ",student_name)
print("Attendance Percentage :  ",Attendance_Percentage)
print("Marks of all subject : ", marks)
print("Final Eligibility status : ",Eligibility_status)
```
